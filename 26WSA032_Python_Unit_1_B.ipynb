# **Structured Programming**

Once you understand how to work with variables and operators, the next step is learning how to combine them into sequences of instructions that a computer will follow step by step - simply writing one statement after another. You have already seen this in many examples above.

However, as programmers we eventually will want to control the flow of the program. This is where control structures come in. With decision-making (using if statements), the program can choose different paths depending on conditions.

With loops (such as for loops), the program can repeat a sequence of instructions many times without you having to write them out one by one. Finally, by creating functions, the programmer can group instructions together in a logical way into reusable blocks that make your code clearer, easier to manage, and more powerful. These building blocks—sequences, decisions, loops, and functions—form the foundation of structured programming.

## **Programming Paradigms**

> Software Engineering

Structured programming belongs to the imperative [programming paradigm](https://en.wikipedia.org/wiki/Programming_paradigm).

In spoken language grammar [the imperative mood is used to form a command or a request ](https://en.wikipedia.org/wiki/Imperative_mood). In imperative software languages we use statements - commands - each of which uses an instruction to the interpreter to act in precise ways on variables using operators.

## **Examples of Structured Programming**

This is an extension of the imperative programming paradigm which makes extensive use of **structured flow constructs** to organise and control the execution of our statements:

* **Sequence** - ordered block of code statements executed in sequence
* **Selection** - use of decision (aka ```if```) to act on conditional or logic
* **Loop** - aka _iteration_ which repeatedly cycles through a sequence block.
* **Functions** - create a sequence which we can call to execute and return results
 statements and then elect to execute a sequence block or an alternative block if given.

 Run the following code cells as examples of each of these.

````python
#@title sequence

import math

radius_inner = 5
volume_inner = 4 / 3 * math.pi * radius_inner**3
radius_outer = 10
volume_outer = 4 / 3 * math.pi * radius_outer**3

net_volume = volume_outer - volume_inner

print ("volume of material in the ball: ", net_volume)
````

````python

#@title decision

import math

radius_inner = 9
radius_outer = 10

if radius_inner > radius_outer:
  print ("volume of material in the ball cannot be calculated if inner radius is larger than the outer radius")
else:
  volume_inner = 4 / 3 * math.pi * radius_inner**3
  volume_outer = 4 / 3 * math.pi * radius_outer**3
  net_volume = volume_outer - volume_inner
  print ("volume of material in the ball: ", net_volume)
````

### **else statement**

This allows the execution of statements if the expression is false. After those are executed, the program flow returns to the current level of indentation of th if statement.

### **elif**
Finally the ```elif``` statement allows a subsequent condition to be tested, if the first has failed. If the first if, and subsequent elif are false, the flow can return to the an else block, or return to the main level following the if statement.

![elif](https://github.com/wattcommunity/Wolfson-Images/blob/main/if_elif_else_flow.jpg?raw=true)

````python
#assign variables
age = 9
height = 1.2
with_parent = True

#evaluate the truth of the expression in an if statement
if age>=12 and height>=1.2:
  print ("customer is allowed to ride")
  print ("Gate opening")
elif age >= 8 and height >= 1.2 and with_parent == True:
  print ("customer is allowed to ride with parent")
  print ("Gate opening")
else:
  print ("customer is not allowed to ride")
  print ("Gate remains closed")

print ("\n")
print ("Next customer please")

````

## **Python range() Function**

There's an answer to this problem. In C++, VB, C# etc we can specifier a loop counter. In python we use the range to generate our indexes in order, and we can then consume each _index_ in a for loop. The next code cell shows how it works, replacing the list of integers used in the previous section.

````python
integers = range(25)

pq = math.pi/4   #prep variables which do not change before the loop

for i in integers:
  np = i*pq
  sin = math.sin(np)
  cos = math.cos(np)

  # note the nice formatting using f-strings
  print (f'{i:4}  {np:10.3f}   {sin:10.3f}   {cos: 10.3f}  {sin**2+ cos**2:10.1f}' )

print ("\nAll printed neatly")

````

Notice I did not print the list returned by range(). What if I put 1,000,000. Would it by that big? What about my computer memory?

Everything is of course safe. The range() call creates an object which is always ready to deliver the next value - in python this is known as a _generator_. Run the next code cell to see (but I do not recommend creating a for loop with the range object created!


````python
#@title demonstrate a very big range
integers = range(99, 200)

i = 0
for integer in integers:
  i += 1
  print (f"{i:5} {integer**0.5:6.3f}")

print ("loop is finished")


````

## **Creating new Lists with a ```for``` Loop and a ```range()```**

As with structures and strings, we can execute a for loop with any size of range. This enables us to create new structures as we itereate through the range.

Lets our sin and cos function this time adding the results of our calculations to a new list object.

> **Walk through the code**


````python
#assign our range generator to a variable 'integers
integers = range(501)              # <- Try different ranges
#Create empty lists for the sin and cos values to which we add our results

sinfunc = []
cosfunc = []
x = []
p = math.pi/250

#Note the use of the list append method
for i in integers:
  np = i*p
  x.append(np)
  sin = math.sin(np)
  cos = math.cos(np)
  sinfunc.append(sin)
  cosfunc.append(round(cos,3))
# At the end of the loop we have three new lists,
# populated with the appended values that have been calculated.
# We can see how long the list have become with len() function

print("\n")
print (f'Length of the list of cosine values is: {len(sinfunc)}')
print (f'Length of the list of sine values is :  {len(cosfunc)}')
print (f'Length of the list of x values is :     {len(x)}')

print (cosfunc)
````

Note how, at each iteration, values are appended to the lists. This will only work if the lists are pre-assigned  to variables before the list starts. The list ```append()``` method is one of many list methods we mentioned . lists and indeed strings, as the above code shows, can also have their length '_measured_' by using the python len() function.
