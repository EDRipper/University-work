## **Definition of a variable**

In programming, a variable is a **named piece of data** that can be stored and used later in a program.

A variable has:

- a symbolic **name** used to refer to it
- a **value** that it currently contains

The value may be a number, text, a logical value (`True` or `False`), or a more complex object.

You have already met variables in mathematics, where symbols are used to represent quantities:

> $a=3$  
> $b=2$  
> $\therefore a^b = 9$

Python uses a similar idea. The statement

```python

```

Variables allow programs to store data, perform calculations, and keep track of results.

# **Variable Identity**

So far we have associated a value with variable name. Bu the variable `name` does not hold the value. Python stores values as **objects** in memory. The `id()` function returns a unique identifier for an object during its lifetime.

If two variables refer to the same object, they will have the same `id()` value.

By examining id() values, we can begin to understand that variables are names that refer to objects stored in memory. This helps explain how Python can use memory efficiently and why changing one variable can sometimes affect another variable if they both refer to the same object.

````python
#@title Using id()

a = 42
b = a
c = 99

print("a =", a, ", id =", id(a))
print("b =", b, ", id =", id(b))
print("c =", c, ", id =", id(c))

````

## **Definition of an operator**
In computer science an operator is a character or characters that instruct the program to perform an **operation** on one or more variables. The types of operation are :

* Arithmetic
* Assignment
* Comparison
* Logical


You will have met some of these operator types in mathematics. **Arithmetic** operators perform the operations of addition, subtraction, multiplication, and division. **Assignment** lets the value on the left of the equals sign be the value returned by the right hand of the equals sign (see below) **Comparison** operators do just that, testing whether two values are the same value, or whether one is greater than or less than the other, for example. **Logical** operators allow us to derive the result of logic operations which are performed with binary variables in ***boolean algebra***.

All the crucial things software is desgined to do - maniulpate and process data, compare and make logical decisions we do with operators. There are other operator types we will come to ad hoc. Lets focus on the key ones for mathematics, which we need to get programming.

## **Arithmetic Operators**

Arithmetic operators are those which perform all the standard mathematical operations: addition, subtraction, multiplication and division. They use the expected symbols ``` + - * / ```. You can probably guess why we use ```*``` not ```x```.

| Operator | Name           | Example     |
|----------|----------------|-------------|

| \*\*     | Exponentiation | z = x ** y  |

Exponentiation - raising one number to the power of another e.g. $a^b$ was touched on above.

Floor division and Modulo$^1$ - you might not have met before, and if you have, not used them very much. The former is a division operator but returns the result without the remainder. The latter returns only the remainder as a result. For example:

> ```5/2 = 2.5```  whereas
>
> ```5//2 = 2```   and
>                    
>  ```5%2 = 1```

Whilst ordinary arithmetic operators are ubiquitous when coding mathematical formulae, floor and modulo operators are suprisingly useful in many programming _patterns_.

## **Assignment Operators**
We saw an assignment operator already. We use it to assigned a value to a variable:

```voltage1 = adc_pin4```

The **assignment** operator used here the ```=``` sign.

On the face of it the assignment operator is at odds with how we understand the symbol in mathematics. It does not so much mean '**equals**', but '**let the value represented by the variable equal the right hand value**'

Consider this assignment
```
x = x + 1
```
In mathematics and algebra we would be saying 0 = 1 which is clearly incorrect. However, in programming we are are saying let x equal the value in x and add 1 to it. This is an important distinction; in mathematics ```=``` is an assertion of equality whereas in programming it is definitively _an assignment_.

## **Compound Assignment Operators**

We can shorten simple assignments such as ```x = x + 1``` using the compound assignment
```
x += 1
```

All the arithmetic operators can be combined with the simple equals assignment operator:

| Operator  | Name           | Example |

| -=        | Subtraction    | x -= y   |
| \*=       | Multiplication | x *= y   |

| \*\*=     | Exponentiation | x **= y  |
| //=       | Floor division | x //= y  |
| %=        | Modulo         | x %= y   |

#### **Logical Operators**

Logical operators can combine boolean values, or expressions which result in boolean values, to determine the resultant  '*truth*' values according to the rules of logic.

> *Which expressions result in boolean values*

| Operator | Description                                             | Example               |
|----------|---------------------------------------------------------|-----------------------|
| and      | Returns True if both statements are true                | x < 5 and  x < 10     |
| or       | Returns True if one of the statements is true           | x < 5 or x < 4        |
| not      | Reverse the result, returns False if the result is true | not(x < 5 and x < 10) |


Therefore we can combine the result of expressions which deliver a boolean result. Consider this statement about Nita and her wellington boots (wellies):

**Statement: Nita is 1.9m tall and has red wellies**

**Are these statements True or False:**

1. Nita is shorter than 2m and has red wellies: True

1. Nita is not shorter than 2m and has red wellies: False

1. Nita is shorter than 2m or has red wellies: True

1. Nita is not shorter than 2m or has red wellies:  True

1. Nita is not shorter than 2m and does not have red wellies: False

1. Nita is shorter than 2m or has red wellies, but not both!: False
