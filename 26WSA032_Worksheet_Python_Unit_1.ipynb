There are 16 ounces in a pound. And there are 14 pounds in a stone.

If a person weighs 2360 ounces, calculate, or convert, using the modulo and floor operators the persons weight in to stones, pounds and ounces.

How to do it

* convert the ounces to pounds by dividing using the modulo operator by 16 to give us the remainder ounces and the floor operator to give the number of pounds.
* Do the same to get the remainer pounds and the number of stones, carefully using 14 as your divisor.
* Add a line of code to verify your answer
* Format the output as string giving the units for each figure.

If you use an LLM to help, make sure you really understand what modulo and floor operators are doing by carefully **walking** through your code to help you follow the code

> **Reflection** why are you asked to **verify** your answer?

````python
#@title answer  exercise 8

total_ounces = 2360  # ounces
stones = total_ounces // (16*14)
pounds = (total_ounces - (stones * 16 * 14)) // 16
ounces = total_ounces - (stones * 16 * 14) - (pounds * 16)
print("Stones:", stones)
print("Pounds:", pounds)
print("Ounces:", ounces)

verification = (stones * 16 * 14) + (pounds * 16) + ounces
print("Verification:", verification)

#im asked to verify because its easier to verify an answer than to work it out from scratch, and it would be an easy check for any errors i missed
````

Our solution is going to calculate the resistance of n resistors in parallel which are passed in a list as an argument to a function we will call ```parallel_resistance```.

Part 2
Take a copy of your function and use it to create a second one
1. Name this new function called resistance_calculator.
2. Have as well as the list of resistor values, pass a string argument `arrangement`
3. If `arrangement` is `'series'` then calculate the total for a series arrangement, else if it is `'parallel'` calculate it for that arrangement.
4. return the result
5. Verify your function
6. Update the doc string








````python
#@title answer 10 - a function for parallel resistors

resistors = [1, 2, 3, 4, 5]
def parallel_resistance(resistors):
    """A function that calculates the total resistance from a list of resistors in parallel, returning the result as a float."""
    sum_of_reciprocals = 0
    for r in resistors:
        sum_of_reciprocals += 1/r
    result = 1/sum_of_reciprocals
    return result

print(parallel_resistance(resistors))

print(parallel_resistance.__doc__)

def resistance(resistors, arrangement):
    """A function that calculates the total resistance from a list of resistors in in series or parallel, returning the result as a float."""
    if arrangement == "series":
        result = sum(resistors)
    elif arrangement == "parallel":
        sum_of_reciprocals = 0
        for r in resistors:
            sum_of_reciprocals += 1/r
        result = 1/sum_of_reciprocals
    else:
        print("Arrangement must be 'series' or 'parallel'")
    return result

print(resistance(resistors, "parallel"))
````
