````python
#@title 9. Avoiding Division by Zero in a Loop
'''
Write your code in this cell to answer exercise 9
'''
import math

numbers = [1,2,0,4,5,7,8,9,0,3,6,7]
answers1 = []
for n in numbers:
  if n == 0:
    print("skipped")
    continue
  else:
    answer = n / (1 - math.exp(n))
    answers1.append(round(answer, 3))
print(answers1)
````
